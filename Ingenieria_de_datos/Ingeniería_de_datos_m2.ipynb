## Preprocesamiento y homologación de la Muestra 2

Con el objetivo de estandarizar la información y facilitar la comparación de los registros de la **Muestra 2** con el catálogo vehicular, se realizó un proceso de homologación de variables orientado a uniformar codificaciones, identificar diferencias de nomenclatura y corregir inconsistencias.

Para preservar la información original, las transformaciones se efectuaron sobre copias de ambas bases de datos, denominadas `muestra_2_homologada` y `catalogo_homologado`, siguiendo el orden que se describe a continuación.

### 1. Homologación de ocupantes

Inicialmente, se estandarizó la variable `Ocupantes`, cuyos valores se encontraban representados mediante cadenas de texto que incluían la capacidad numérica del vehículo y la abreviatura `OCUP.`.

Se extrajo la cantidad de ocupantes y se convirtió a formato numérico entero, eliminando los ceros iniciales y el texto adicional.

| Formato original | Formato homologado |
|---|---|
| 02 OCUP. | 2 |
| 04 OCUP. | 4 |
| 05 OCUP. | 5 |
| 07 OCUP. | 7 |

Esta transformación permitió unificar la representación de la capacidad de ocupantes con el formato utilizado en el catálogo vehicular. Posteriormente, se examinó la distribución de los valores homologados.

### 2. Revisión de marcas

Se compararon los valores únicos de la variable `Marca` con las denominaciones disponibles en el catálogo vehicular, identificando las marcas coincidentes y aquellas sin correspondencia exacta.

A diferencia de la Muestra 1, en esta etapa no se establecieron sustituciones específicas para las marcas de la Muestra 2, por lo que todas las denominaciones se conservaron sin modificaciones.

### 3. Homologación de aire acondicionado y quemacocos

Se estandarizó la codificación de las variables `AC` y `Quemacocos`, originalmente representadas mediante valores binarios (0 y 1), utilizando la nomenclatura del catálogo vehicular.

| Variable | Código original | Código homologado | Significado |
|---|---|---|---|
| AC | 0 | SA | Sin aire acondicionado |
| AC | 1 | CA | Con aire acondicionado |
| Quemacocos | 0 | SQ | Sin quemacocos |
| Quemacocos | 1 | CQ | Con quemacocos |

Estas sustituciones permitieron uniformar la representación de ambas características sin alterar su significado.

### 4. Limpieza de descripciones y corrección de transmisión

La siguiente etapa consistió en revisar la información contenida en `descripcion` para identificar posibles inconsistencias en la variable `transmision`.

Se utilizaron las terminaciones `AUT`, correspondiente a **transmisión automática**, y `STD`, correspondiente a **transmisión estándar o manual**, como referencias para comprobar la consistencia de los registros.

El procedimiento se desarrolló en el siguiente orden:

1. **Limpieza de descripciones:** se eliminaron los espacios sobrantes al inicio y al final de las cadenas de texto, así como los puntos ubicados al final de las descripciones. Adicionalmente, se eliminaron los espacios que pudieran permanecer después de retirar los puntos, corrigiendo terminaciones como `AUT .` y `STD .`.
2. **Identificación de transmisión:** se extrajo la última palabra de cada descripción para identificar las terminaciones `AUT` y `STD`.
3. **Detección de discrepancias:** se comparó la terminación extraída con el valor registrado en `transmision`, considerando únicamente las descripciones con terminaciones válidas.
4. **Corrección de inconsistencias:** cuando se detectó una discrepancia, se reemplazó el valor de `transmision` por el indicado en la descripción.
5. **Verificación:** se contabilizaron las transmisiones corregidas y se comprobó la cantidad de descripciones sin terminación válida.

### 5. Homologación de transmisión

Una vez concluida la revisión y corrección de inconsistencias, se transformaron las categorías de `transmision` a la nomenclatura utilizada en el catálogo vehicular:

| Código original | Código homologado | Significado |
|---|---|---|
| AUT | A | Transmisión automática |
| STD | S | Transmisión estándar o manual |

Esta homologación se efectuó después de las correcciones basadas en la descripción, dado que estas utilizaron las categorías originales como referencia. Posteriormente, se examinó la distribución de las transmisiones homologadas.

### 6. Homologación de motor en el catálogo vehicular

Posteriormente, se revisó la variable `MOTOR` del catálogo vehicular, identificando diferencias en el uso de letras mayúsculas y minúsculas para representar las configuraciones del motor.

Para uniformar su nomenclatura, se convirtieron las cadenas de texto a mayúsculas, conservando su significado original.

| Denominación original | Denominación homologada |
|---|---|
| l4 | L4 |
| v6 | V6 |
| v8 | V8 |


Esta transformación se aplicó exclusivamente sobre la copia del catálogo vehicular, sin modificar la variable `motor` de la Muestra 2. Los valores faltantes del catálogo se conservaron sin cambios.

Finalmente, se examinó la distribución de las configuraciones de motor, incluyendo los valores faltantes.

### 7. Revisión de la variable Tipo

Finalmente, se compararon las denominaciones de la variable `Tipo` con las existentes en el catálogo vehicular. Para ello, se excluyeron los valores faltantes y se convirtieron temporalmente los valores a cadenas de texto, permitiendo incluir las categorías originalmente representadas mediante valores numéricos.

Asimismo, se eliminaron los espacios al inicio y al final de las cadenas utilizadas en la comparación.

Se identificaron las categorías coincidentes y aquellas presentes exclusivamente en cada base de datos. No se modificaron los valores originales de esta variable, ya que sus diferencias de nomenclatura serán abordadas posteriormente mediante medidas de disimilitud de cadenas de caracteres.

### Alcance de las transformaciones

Las homologaciones se limitaron a las equivalencias previamente identificadas y a las inconsistencias que podían corregirse mediante la información disponible. En particular, la conversión de `Ocupantes` presupone que los registros contienen una cantidad numérica reconocible con el formato esperado; los valores que no cumplan esta condición requieren revisión antes de completar la transformación.

Por otra parte, las correcciones de transmisión se realizaron únicamente cuando la descripción proporcionaba una terminación válida. No se efectuaron procedimientos generales de imputación ni eliminación de registros por valores faltantes. Estos se conservaron en las operaciones que lo permitían, mientras que en la comparación de `Tipo` se excluyeron únicamente de los conjuntos utilizados para el análisis.

Por lo tanto, el procedimiento **no garantiza la correspondencia exacta de todos los registros con el catálogo vehicular**, sino que constituye una etapa de preparación y estandarización de los datos para su posterior comparación y asignación de claves.
