## Preprocesamiento y homologación de la Muestra 1

Con el objetivo de estandarizar la información y facilitar la comparación de los registros de la **Muestra 1** con el catálogo vehicular, se realizó un proceso de homologación de variables orientado a uniformar codificaciones, identificar diferencias de nomenclatura y corregir inconsistencias.

Para preservar la información original, las transformaciones se efectuaron sobre una copia de la base de datos, denominada `muestra_1_homologada`, siguiendo el orden que se describe a continuación.

### 1. Homologación de marcas

Inicialmente, se identificaron abreviaturas de la variable `Marca` cuya equivalencia con las denominaciones del catálogo podía establecerse directamente. Se realizaron las siguientes sustituciones:

| Código original | Denominación homologada |
|---|---|
| MBENZ | MERCEDES BENZ |
| LROVER | LAND ROVER |

Posteriormente, se compararon los valores únicos de ambas bases de datos para identificar las marcas coincidentes y aquellas sin correspondencia exacta. Las denominaciones restantes se conservaron sin modificaciones.

### 2. Homologación de aire acondicionado y quemacocos

Se estandarizó la codificación de las variables `AC` y `Quemacocos`, originalmente representadas mediante valores binarios (0 y 1), utilizando la nomenclatura del catálogo vehicular.

| Variable | Código original | Código homologado | Significado |
|---|---|---|---|
| AC | 0 | SA | Sin aire acondicionado |
| AC | 1 | CA | Con aire acondicionado |
| Quemacocos | 0 | SQ | Sin quemacocos |
| Quemacocos | 1 | CQ | Con quemacocos |

Estas sustituciones permitieron uniformar la representación de ambas características sin alterar su significado.

### 3. Limpieza de descripciones y corrección de transmisión

La siguiente etapa consistió en revisar la información contenida en `descripcion` para identificar posibles inconsistencias en la variable `transmision`.

Se utilizaron las terminaciones `AUT`, correspondiente a **transmisión automática**, y `STD`, correspondiente a **transmisión estándar o manual**, como referencias para comprobar la consistencia de los registros.

El procedimiento se desarrolló en el siguiente orden:

1. **Limpieza de descripciones:** se eliminaron los espacios sobrantes al inicio y al final de las cadenas de texto, así como los puntos ubicados al final de las descripciones.
2. **Identificación de transmisión:** se extrajo la última palabra de cada descripción para identificar las terminaciones `AUT` y `STD`.
3. **Detección de discrepancias:** se comparó la terminación extraída con el valor registrado en `transmision`, considerando únicamente las descripciones con terminaciones válidas.
4. **Corrección de inconsistencias:** cuando se detectó una discrepancia, se reemplazó el valor de `transmision` por el indicado en la descripción.
5. **Verificación:** se contabilizaron las transmisiones corregidas, se examinó la distribución de las categorías resultantes y se comprobó la cantidad de descripciones sin terminación válida.

### 4. Homologación de transmisión

Una vez concluida la revisión y corrección de inconsistencias, se transformaron las categorías de `transmision` a la nomenclatura utilizada en el catálogo vehicular:

| Código original | Código homologado | Significado |
|---|---|---|
| AUT | A | Transmisión automática |
| STD | S | Transmisión estándar o manual |

Esta homologación se efectuó después de las correcciones basadas en la descripción, dado que estas utilizaron las categorías originales como referencia.

### 5. Revisión de la variable Tipo

Finalmente, se compararon las denominaciones de la variable `Tipo` con las existentes en el catálogo vehicular. Para ello, se excluyeron los valores faltantes y se eliminaron los espacios al inicio y al final de las cadenas utilizadas en la comparación.

Se identificaron las categorías coincidentes y aquellas presentes exclusivamente en cada base de datos. No se modificaron los valores originales de esta variable, ya que sus diferencias de nomenclatura serán abordadas posteriormente mediante medidas de disimilitud de cadenas de caracteres.

### Alcance de las transformaciones

Las homologaciones se limitaron a las equivalencias previamente identificadas y a las inconsistencias que podían corregirse mediante la información disponible. En particular, las correcciones de transmisión se realizaron únicamente cuando la descripción proporcionaba una terminación válida, mientras que las categorías sin una regla de transformación definida conservaron sus valores originales.

Por lo tanto, el procedimiento **no garantiza la correspondencia exacta de todos los registros con el catálogo vehicular**, sino que constituye una etapa de preparación y estandarización de los datos para su posterior comparación y asignación de claves.
