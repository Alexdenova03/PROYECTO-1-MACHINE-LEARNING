### Variables binarias

Se analizarán las variables `AC`, `Quemacocos`, `TURBO` y `hibrido`, mediante sus frecuencias absolutas y relativas, con el objetivo de identificar la distribución de sus categorías y posibles concentraciones en los datos.

```python
variables_binarias = ["AC", "Quemacocos", "TURBO", "hibrido"]

for variable in variables_binarias:
    frecuencias = muestra_1[variable].value_counts().sort_index()

    tabla = pd.DataFrame({
        "Frecuencia absoluta": frecuencias,
        "Frecuencia relativa (%)": (frecuencias / len(muestra_1) * 100).round(2)
    })

    display(tabla)
```

En las cuatro variables binarias se observa una mayor frecuencia de la categoría 0. Las variables `AC` y `TURBO` presentan distribuciones relativamente similares, con aproximadamente una cuarta parte de los vehículos registrados con estas características (24.93% y 26.60%, respectivamente). En contraste, la presencia de quemacocos es menos frecuente (6.38%), mientras que únicamente seis vehículos (0.21%) están identificados como híbridos. Por tanto, `Quemacocos` e `hibrido` presentan una concentración especialmente elevada en la categoría 0.

### Variables cuantitativas

Se analizarán las variables `Modelo` y `cilindrada` mediante estadísticos descriptivos y distribuciones de frecuencias. El objetivo es caracterizar sus valores, identificar su tendencia central y dispersión, así como examinar posibles concentraciones en la muestra.

#### Modelo

La variable `Modelo` representa el año-modelo del vehículo y no presenta valores faltantes, por lo que se analizarán las 2,868 observaciones.

Se calcularán el mínimo, máximo, media, mediana, moda, cuartiles y desviación estándar. Adicionalmente, se obtendrán las frecuencias absolutas y relativas de cada año-modelo, con el propósito de identificar los años más frecuentes y la distribución temporal de los vehículos.

#### Modelo

```python

modelo = muestra_1["Modelo"]

# Estadísticos descriptivos del año-modelo
resumen_modelo = pd.Series({
    "Mínimo": modelo.min(),
    "Q1": modelo.quantile(0.25),
    "Mediana": modelo.median(),
    "Media": round(modelo.mean(), 2),
    "Q3": modelo.quantile(0.75),
    "Máximo": modelo.max(),
    "Moda": modelo.mode().tolist(),
    "Desviación estándar": round(modelo.std(), 2)
})

display(resumen_modelo.to_frame("Valor"))

# Distribución de frecuencias por año-modelo
frecuencias = modelo.value_counts().sort_index()

tabla_modelo = pd.DataFrame({
    "Frecuencia absoluta": frecuencias,
    "Frecuencia relativa (%)": (frecuencias / len(modelo) * 100).round(2)
})

display(tabla_modelo)
```

La variable `Modelo` presenta valores comprendidos entre 1978 y 2020, con una media de 2008.53 y una mediana de 2010. El 50% central de las observaciones se encuentra aproximadamente entre los años-modelo 2004 y 2015, mientras que la desviación estándar es de 7.88 años.

El año-modelo más frecuente es 2014, con 197 vehículos (6.87%). La distribución muestra una mayor concentración en los años más recientes del intervalo observado, mientras que los vehículos correspondientes a las décadas de 1970, 1980 y 1990 presentan frecuencias considerablemente menores.

#### Cilindrada


```python

cilindrada = muestra_1["cilindrada"]
cilindrada_validos = cilindrada.dropna()

# Estadísticos descriptivos de cilindrada
resumen_cilindrada = pd.Series({
    "Observaciones válidas": len(cilindrada_validos),
    "Valores faltantes": cilindrada.isna().sum(),
    "Mínimo": cilindrada_validos.min(),
    "Q1": cilindrada_validos.quantile(0.25),
    "Mediana": cilindrada_validos.median(),
    "Media": round(cilindrada_validos.mean(), 2),
    "Q3": cilindrada_validos.quantile(0.75),
    "Máximo": cilindrada_validos.max(),
    "Moda": cilindrada_validos.mode().tolist(),
    "Desviación estándar": round(cilindrada_validos.std(), 2)
})

display(resumen_cilindrada.to_frame("Valor"))

# Frecuencias sobre las 2,868 observaciones
frecuencias = cilindrada.value_counts(dropna=False).sort_index()

tabla_cilindrada = pd.DataFrame({
    "Frecuencia absoluta": frecuencias,
    "Frecuencia relativa (%)": (frecuencias / len(cilindrada) * 100).round(2)
})

tabla_cilindrada.index = tabla_cilindrada.index.fillna("Sin dato")
display(tabla_cilindrada)

# Frecuencias únicamente entre los registros disponibles
frecuencias_validas = cilindrada_validos.value_counts().sort_index()

tabla_cilindrada_validos = pd.DataFrame({
    "Frecuencia absoluta": frecuencias_validas,
    "Frecuencia relativa (%)": (
        frecuencias_validas / len(cilindrada_validos) * 100
    ).round(2)
})

display(tabla_cilindrada_validos)
```

La variable `cilindrada` presenta 1,821 valores faltantes (63.49%), por lo que los estadísticos descriptivos se calcularon únicamente sobre las 1,047 observaciones disponibles.

Los valores registrados se encuentran entre 1.0 y 6.4 litros, con una media de 2.8 litros y una mediana de 2.1 litros. El 50% central de las observaciones se sitúa entre 1.8 y 3.6 litros, con una desviación estándar de 1.32 litros.

La cilindrada más frecuente es 2.0 litros, presente en 229 vehículos (21.87% de los registros disponibles), seguida de 3.0 litros, con 141 vehículos (13.47%). La diferencia entre la media y la mediana sugiere una distribución con cierta asimetría hacia valores elevados.

Finalmente, la elevada proporción de valores faltantes limita la representatividad de estos resultados respecto al conjunto de la muestra y deberá considerarse durante la ingeniería de datos.

### Variable cualitativa textual

#### descripcion

La variable `descripcion` contiene información comercial y técnica de los vehículos mediante cadenas de caracteres. Presenta 1,737 valores distintos entre las 2,868 observaciones de la muestra y no contiene valores faltantes.

Debido a su elevada cardinalidad, el análisis se centrará en estudiar la diversidad y repetición de las descripciones. Para ello, se calculará la proporción de descripciones distintas respecto al total de observaciones, así como la cantidad y proporción de descripciones que aparecen una sola vez o múltiples veces.

Adicionalmente, se distinguirá entre el porcentaje de descripciones distintas que se repiten y el porcentaje de observaciones que contienen descripciones repetidas.

```python


descripcion = muestra_1["descripcion"]
frecuencias = descripcion.value_counts()

total = len(descripcion)
distintas = len(frecuencias)
unicas = (frecuencias == 1).sum()
repetidas = (frecuencias > 1).sum()
observaciones_repetidas = frecuencias[frecuencias > 1].sum()

tabla_descripcion = pd.DataFrame({
    "Indicador": [
        "Descripciones distintas",
        "Descripciones que aparecen una sola vez",
        "Descripciones que aparecen más de una vez",
        "Observaciones con descripción única",
        "Observaciones con descripción repetida"
    ],
    "Frecuencia absoluta": [
        distintas, unicas, repetidas, unicas, observaciones_repetidas
    ],
    "Frecuencia relativa (%)": [
        distintas / total * 100,
        unicas / distintas * 100,
        repetidas / distintas * 100,
        unicas / total * 100,
        observaciones_repetidas / total * 100
    ],
    "Base del porcentaje": [
        "Total de observaciones",
        "Descripciones distintas",
        "Descripciones distintas",
        "Total de observaciones",
        "Total de observaciones"
    ]
})

tabla_descripcion["Frecuencia relativa (%)"] = (
    tabla_descripcion["Frecuencia relativa (%)"].round(2)
)

display(tabla_descripcion)

```
