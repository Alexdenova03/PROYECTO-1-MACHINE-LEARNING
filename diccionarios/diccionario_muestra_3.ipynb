# Diccionario de datos — Muestra 3

La Muestra 3 contiene **803 observaciones y 3 variables** que describen características de los vehículos.

El presente diccionario tiene como objetivo documentar las características de cada variable, incluyendo su significado, tipo de dato, naturaleza estadística, valores observados y las anomalías estructurales de captura identificadas durante la exploración.

## Descripción de las variables

### 1. `descripcion`

- **Descripción:** contiene una cadena de caracteres que reúne información comercial y técnica del vehículo, incluyendo su marca, denominación comercial y, dependiendo del registro, características mecánicas, de carrocería, equipamiento, transmisión y, en un subconjunto de observaciones, la capacidad de ocupantes.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa textual.
- **Valores distintos:** 508.
- **Valores faltantes:** 0 (0.00%).

**Análisis de la estructura de las descripciones**

Debido a la elevada cardinalidad de la variable, se realizó una inspección de las diez descripciones más frecuentes y de una selección aleatoria de 40 descripciones distintas, excluyendo las diez anteriores.

A partir de esta revisión, se identificó la siguiente estructura general en la mayoría de los registros:

**Marca o prefijo → Denominación comercial → Características adicionales → Transmisión (y, en casos con desfase, Ocupantes).**

Las características adicionales pueden incluir:

- **Versión o edición:** denominaciones particulares del vehículo, como `CARRERA 4S`, `PLATINUM`, `LAREDO`, `PRESTIGE`, `R-DYNAMIC`, `WRX STI` o `TRACKHAWK`.
- **Carrocería y puertas:** configuraciones como `CONVERTIBLE`, `COUPE`, `CABRIO`, `TARGA`, `SEDAN`, `HATCHBACK`, `SUV` o `SPORT TURISMO`, así como el conteo de puertas (`2P`, `4P`, `5P`).
- **Motor y sobrealimentación:** configuración de cilindros (`L4`, `L6`, `V6`, `V8`, `B4` [bóxer de 4 cilindros], `H4`, `H6` [horizontalmente opuestos]), cilindrada (`1.4L`, `2.0L`, `2.9L`, `3.0L`, `3.8L`, `5.0L`, `6.2L`), potencia (`265HP`, `300CP`, `340CP`, `423HP`) y sistemas de inducción o propulsión (`TURBO`, `BITURBO`, `SC` [supercargador], `GTDI`, `TSI`, `HIBRIDO`, `ELECTRICO`).
- **Tracción:** denominaciones como `4X2`, `4X4`, `AWD` y `FWD`.
- **Equipamiento:** características como frenos antibloqueo (`ABS`), bolsas de aire (`BA`), aire acondicionado (`AC`), quemacocos o techo panorámico (`QC`, `Q/C`, `TPANOR`, `TCORR`, `TECHO LONA`, `TECHO DURO`), vestiduras de piel (`VP`, `V/P`, `LEATHERETTE`), navegación (`NAVI`) y diámetro de rin (`R16`, `R17`, `R18`, `R19`).
- **Transmisión:** denominaciones estándar como `AUT.` y `STD.`, o sistemas específicos como `PDK`, `PDK-7V`, `TIPTRONIC-8V`, `CVT`, `SELECTSHIFT`, `6VEL` y `7SPEED`.

La cantidad y el orden de los elementos intermedios no son uniformes entre las descripciones. Mientras que algunos registros contienen únicamente la marca, el modelo comercial y la transmisión (por ejemplo, `LINCOLN LS AUT.` o `JAGUAR XJ8 AUT.`), otros detallan exhaustivamente el tren motriz y el equipamiento.

**Análisis de equivalencias y relaciones comerciales de las marcas identificadas**

La inspección de las 15 marcas presentes en la muestra muestra un perfil de cartera fuertemente concentrado en vehículos deportivos, de lujo y SUVs, entre los cuales existen vínculos corporativos e industriales relevantes:

1. **Equivalencias de denominación y variantes ortográficas en modelos:**
   - `AR ALFA ROMEO` y `ALFA ROMEO`: identifican a Alfa Romeo.
   - `LX LEXUS` y `LEXUS`: identifican a Lexus.
   - En los modelos de Porsche se detectó una inconsistencia ortográfica frecuente entre `PORSCHE BOXSTER` (12 registros, grafía correcta) y `PORSCHE BOXTER` (11 registros, omisión de la letra *S*), lo cual es relevante para las métricas de similitud léxica.

2. **Relaciones empresariales vigentes en 2026:**
   - **Porsche, Lamborghini y Cupra:** las tres marcas pertenecen a **Volkswagen Group**. Porsche y Lamborghini integran el grupo de marcas deportivas y de superlujo, mientras que Cupra se independizó en 2018 como la división deportiva de SEAT dentro del mismo consorcio.
   - **Jeep y Alfa Romeo:** ambas pertenecen a **Stellantis**, conglomerado formado en 2021 tras la fusión de Fiat Chrysler Automobiles (FCA) y Groupe PSA.
   - **Cadillac y Buick:** ambas son divisiones históricas vigentes pertenecientes a **General Motors (GM)**.
   - **Lincoln:** constituye la división de vehículos de lujo de **Ford Motor Company** desde su adquisición en 1922.
   - **Infiniti:** es la división de vehículos de lujo de **Nissan Motor Corporation**, creada en 1989.
   - **Lexus:** es la división de vehículos de lujo de **Toyota Motor Corporation**, fundada en 1989; Asimismo, Toyota mantiene una participación accionaria e industrial en **Subaru Corporation**.
   - **Jaguar:** pertenece al grupo británico **JLR (Jaguar Land Rover)**, el cual es propiedad del conglomerado indio **Tata Motors** desde 2008.
   - **Lotus:** pertenece mayoritariamente desde 2017 al grupo automotriz chino **Zhejiang Geely Holding Group**.
   - **SsangYong:** fabricante surcoreano que en 2022 fue adquirido por KG Group y renombrado corporativamente como **KG Mobility (KGM)** en 2023.
   - **JAC:** fabricante chino (*Anhui Jianghuai Automobile Group*) cuyos vehículos son ensamblados y comercializados en México en alianza con **Giant Motors Latinoamérica**.

3. **Relaciones empresariales históricas:**
   - **Jaguar y Lincoln (Ford Premier Automotive Group):** Jaguar perteneció a Ford Motor Company entre 1989 y 2008, compartiendo plataformas y motores (como los bloques V6 y V8 de los modelos *Jaguar S-Type*, *XJ8* y *Lincoln LS* presentes en la muestra) antes de su venta a Tata Motors.
   - **Lotus y General Motors:** Lotus perteneció a General Motors entre 1986 y 1993, y posteriormente al fabricante malasio Proton entre 1996 y 2017.

**Identificación de las terminaciones y anomalías en `descripcion`**

Al verificar el cierre de las 803 cadenas de texto, se identificaron tres patrones distintos que explican el comportamiento del resto de las columnas:

- **Terminación estándar con transmisión (736 registros, 91.53%):** 550 descripciones finalizan con `AUT.` (68.49%), 185 finalizan con `STD.` (23.04%) y 1 finaliza con `AWD.` (0.12%).
- **Terminación con capacidad de ocupantes incrustada (60 registros, 7.47%):** 58 descripciones finalizan con el token `OCUP` precedido del número de plazas (ej. `... AUT 5 OCUP` o `... PDK 4 OCUP`), además de 2 registros de la misma estructura que cierran con `R19` o `.` tras un desfase de captura. En todos estos casos, el atributo de ocupantes fue absorbido dentro de `descripcion`.
- **Terminación truncada antes de la transmisión (7 registros, 0.87%):** 4 descripciones terminan en `BITURBO`, 2 en `TURBO` y 1 en `5P`. En estos registros, la parte final de la descripción (la transmisión) se desbordó hacia la columna contigua `Ocupantes`.

### 2. `Ocupantes`

- **Descripción:** representa la capacidad nominal de pasajeros (plazas) del vehículo. Sin embargo, debido a errores de alineación en la fuente de datos, en un subconjunto de registros almacena el año-modelo o el tipo de transmisión.
- **Tipo de dato original:** `object` (combina internamente cadenas `str` y enteros `int` en Python).
- **Naturaleza de la variable:** cuantitativa discreta en su sentido teórico (conteo de plazas), pero almacenada como cualitativa nominal / mixta debido al formato de texto (`"05 OCUP."`) y a la contaminación cruzada de columnas.
- **Valores distintos:** 13.
- **Valores faltantes explícitos (`NaN`):** 0 (0.00%).
- **Valores faltantes implícitos (por desplazamiento de columna):** 67 registros (8.34%) no contienen el número de ocupantes en esta columna, aunque en 60 de ellos el dato puede recuperarse del final de `descripcion` y en los 7 restantes se desplazó hacia `Modelo`.

### 3. `Modelo`

- **Descripción:** representa el año-modelo del vehículo. No obstante, en los registros afectados por el desplazamiento de columnas, contiene el nombre de la marca del vehículo o la capacidad de ocupantes.
- **Tipo de dato original:** `object` (combina internamente enteros `int` y cadenas `str` en Python).
- **Naturaleza de la variable:** cuantitativa discreta con interpretación temporal (en los registros válidos), distorsionada a tipo mixto por la infiltración de cadenas de texto.
- **Valores distintos:** 45 en total (29 años numéricos válidos y 16 categorías textuales infiltradas).
- **Intervalo observado de años válidos:** de 1989 a 2020 (nota: no se observan vehículos de los años 1990, 1991 ni 1992 en la muestra).
- **Valores faltantes explícitos (`NaN`):** 0 (0.00%).
- **Valores faltantes implícitos (por pérdida del año-modelo):** 7 registros (0.87%) carecen por completo de año-modelo en toda la fila debido a que el desbordamiento hacia la derecha empujó el año fuera de las tres columnas disponibles.

**Análisis de los valores observados y anomalías en `Modelo`**

1. **Registros numéricos válidos En `Modelo` (736 observaciones, 91.66%):**
   Contienen directamente el año-modelo como entero entre `1989` y `2020` (`[1989, 1993, 1994, ..., 2019, 2020]`).

2. **Registros con marcas infiltradas por desplazamiento a la izquierda (60 observaciones, 7.47%):**
   Cuando el número de ocupantes quedó incrustado al final de `descripcion` y el año-modelo (`2017`–`2020`) pasó a la columna `Ocupantes`, la columna `Modelo` recibió el nombre de la marca del vehículo:
   - `JAGUAR`: 26 registros (12 del año 2019, 9 de 2020 y 5 de 2018).
   - `PORSCHE`: 6 registros (2 de 2020, 2 de 2019 y 2 de 2018).
   - `JEEP`: 6 registros (5 de 2018 y 1 de 2019).
   - `INFINITI`: 6 registros (5 de 2020 y 1 de 2019).
   - `LINCOLN`: 4 registros (2 de 2020, 1 de 2018 y 1 de 2017).
   - `LEXUS`: 3 registros (todos de 2020).
   - `SUBARU`: 2 registros (ambos de 2018).
   - `LAMBORGHINI`: 2 registros (1 de 2020 y 1 de 2018).
   - `CUPRA`, `JAC`, `ALFA ROMEO`, `BUICK` y `CADILLAC`: 1 registro cada una (entre 2019 y 2020).

3. **Registros con ocupantes infiltrados por desplazamiento a la derecha (7 observaciones, 0.87%):**
   Cuando la transmisión se desbordó hacia `Ocupantes` (`PDK AUT.`, `AUT.`, `STD.`), la capacidad de pasajeros fue desplazada hacia `Modelo`:
   - `04 OCUP.`: 4 registros (asociados con `PDK AUT.` en `Ocupantes`).
   - `02 OCUP.`: 2 registros (1 asociado con `PDK AUT.` y 1 con `STD.`).
   - `05 OCUP.`: 1 registro (asociado con `AUT.` en `Ocupantes`).
   En estas 7 observaciones el año-modelo se perdió al desplazarse fuera del límite derecho de la tabla, por lo que constituyen **valores faltantes reales del año-modelo** que deberán tratarse durante la etapa de ingeniería de datos.

## Síntesis del diagnóstico estructural de la Muestra 3

A pesar de que la inspección superficial mediante `.isna().sum()` reporta **0 valores nulos (`NaN`)** en las tres variables, el análisis exhaustivo demuestra que la Muestra 3 presenta tres regímenes de estructura tabular que deben homologarse en la etapa de Ingeniería de Datos antes de ejecutar cualquier algoritmo de claveteo:

| Régimen de captura | Filas afectadas | % de la muestra | Contenido en `descripcion` | Contenido en `Ocupantes` | Contenido en `Modelo` | Acción requerida en Ingeniería de Datos |
|---|---:|---:|---|---|---|---|
| **1. Estándar (Correcto)** | 736 | 91.66% | Descripción + Transmisión | Capacidad (`0X OCUP.`) | Año (`1989`–`2020`) | Limpieza léxica y extracción de entero en `Ocupantes`. |
| **2. Desfase a la izquierda** | 60 | 7.47% | Descripción + Transmisión + `X OCUP` | Año (`2017`–`2020`) | Marca (`JAGUAR`, etc.) | Extraer ocupantes de `descripcion`, mover el año de `Ocupantes` a `Modelo` y conservar la marca. |
| **3. Desfase a la derecha** | 7 | 0.87% | Descripción truncada | Transmisión (`PDK AUT.`, `AUT.`, `STD.`) | Capacidad (`0X OCUP.`) | Concatenar `Ocupantes` al final de `descripcion`, mover `Modelo` a `Ocupantes` y tratar el año como faltante (`NaN`). |
