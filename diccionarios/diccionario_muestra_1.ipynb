# Diccionario de datos — Muestra 1

La Muestra 1 contiene **2,868 observaciones y 12 variables** que describen características de los vehículos.

El presente diccionario tiene como objetivo documentar las características de cada variable, incluyendo su significado, tipo de dato, naturaleza estadística y valores observados.

## Descripción de las variables

### 1. `descripcion`

- **Descripción:** contiene una cadena de caracteres que reúne información comercial y técnica del vehículo, incluyendo su marca, denominación comercial y, dependiendo del registro, características mecánicas, de carrocería y de equipamiento.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa textual.
- **Valores distintos:** 1,737.
- **Valores faltantes:** 0.

**Análisis de la estructura de las descripciones**

Debido a la elevada cardinalidad de la variable, se realizó una inspección de las diez descripciones más frecuentes y de una selección aleatoria de 40 descripciones distintas, excluyendo las diez anteriores.

A partir de esta revisión, se identificó la siguiente estructura general:

**Marca o abreviatura → Denominación comercial → Características adicionales → Transmisión.**

Las características adicionales pueden incluir:

- **Versión o edición:** denominaciones particulares del vehículo, como `SS`, `SLT` o `STYLE`.
- **Carrocería:** características como `SEDAN`, `COUPE`, `CABRIOLET` o `CONVERTIBLE`.
- **Motor:** configuración y cilindrada, como `V6`, `L4`, `2.0L` o `3.5L`.
- **Tracción:** denominaciones como `AWD`, `4X2` o `4MATIC`.
- **Equipamiento:** características como aire acondicionado, quemacocos, bolsas de aire o sistemas de navegación.
- **Transmisión:** denominaciones como `AUT` y `STD`.

La cantidad y el orden de los elementos intermedios no son uniformes entre las descripciones. Algunos registros contienen únicamente la denominación comercial y la transmisión, mientras que otros incluyen numerosas especificaciones adicionales.

**Identificación de las marcas**

Se extrajo la primera palabra de las 2,868 descripciones, identificándose 25 prefijos distintos. Posteriormente, se compararon dichos prefijos con la variable `Marca` para establecer sus correspondencias.

Se encontraron las siguientes equivalencias y asociaciones:

| Prefijo | Marca registrada |
|---|---|
| `AT` | ASTON MARTIN |
| `CR` | CHRYSLER |
| `GIANT` | GIANT MOTORS |
| `GMOTORS` | GENERAL MOTORS / GMOTORS |
| `LROVER` | LROVER (Land Rover) |
| `MBENZ` | MERCEDES BENZ / MBENZ |
| `ML` | MCLAREN |
| `MN` | MINI |
| `MT` | MASERATI |
| `PK` | PONTIAC |
| `TE` | TE TESLA / TESLA MOTORS |
| `VALIANT` | CHRYSLER |

Los 13 prefijos restantes (`ACURA`, `AUDI`, `BAIC`, `BENTLEY`, `BMW`, `CHEVROLET`, `CHRYSLER`, `DODGE`, `FAW`, `HUMMER`, `MAYBACH`, `OLDSMOBILE` y `SAAB`) coinciden directamente con la denominación registrada en `Marca`.

El prefijo `VALIANT` constituye un caso particular, pues corresponde a una denominación comercial y no al nombre de la marca registrada (`CHRYSLER`).

Estas correspondencias fueron identificadas mediante el análisis completo de la muestra y representan asociaciones observadas entre ambas columnas.

**Identificación de la denominación comercial**

Se observó que, después del prefijo inicial, generalmente aparece información que identifica comercialmente al vehículo. Sin embargo, esta información no siempre coincide literalmente con la variable `Tipo`.

Por ejemplo:

- `CHEVROLET CAMARO SS AUT.`: la descripción contiene directamente el nombre comercial Camaro.
- `LROVER RANGE ROVER SPORT...`: la denominación comercial contiene varias palabras.
- `BMW 120IA...`: presenta un código de modelo específico, mientras que `Tipo` registra `SERIE 1`.
- `MBENZ C-63 AMG...`: identifica un modelo específico cuya categoría en `Tipo` es `SERIE C`.

Por tanto, las descripciones presentan diferentes niveles de detalle y emplean nomenclaturas que no necesariamente son idénticas a las de `Tipo`.

**Identificación de la transmisión**

Se verificaron las terminaciones de las 2,868 descripciones mediante una expresión regular, comprobándose que todas finalizan con alguna de las siguientes denominaciones:

- `AUT`: transmisión automática.
- `STD`: transmisión estándar o manual.

Este patrón constituye una característica uniforme de las descripciones examinadas.

**Conclusión del análisis**

La variable `descripcion` contiene información vehicular estructurada parcialmente dentro de una cadena de texto. Aunque se identificaron patrones consistentes en sus elementos iniciales y finales, el contenido intermedio presenta diferencias importantes en cuanto a nomenclatura, cantidad de información y nivel de especificidad.

La inspección exploratoria permitió caracterizar su contenido general, mientras que el análisis de la totalidad de los registros permitió comprobar las correspondencias de los prefijos con `Marca` y la presencia de las terminaciones asociadas con la transmisión.

**Análisis de equivalencias y relaciones comerciales**

La inspección de las categorías revela que no todas representan marcas automotrices diferentes. Algunas corresponden a abreviaturas, denominaciones históricas, grupos empresariales o compañías ensambladoras.

**1. Equivalencias de denominación**

- `MBENZ` y `MERCEDES BENZ`: ambas denominaciones identifican Mercedes-Benz.
- `LROVER`: abreviatura de Land Rover.
- `GENERAL MOTORS` y `GMOTORS`: denominaciones aparentemente equivalentes del grupo General Motors. La identificación de la marca comercial específica requiere información adicional.
- `TE TESLA` y `TESLA MOTORS`: posibles denominaciones de Tesla. Tesla Motors cambió su nombre corporativo a Tesla, Inc. en 2017. El prefijo `TE` requiere confirmación.

**2. Relaciones empresariales vigentes en 2026**

- **BMW y MINI:** ambas marcas pertenecen actualmente a BMW Group. MINI se incorporó al grupo en 1994.
- **Audi y Bentley:** ambas pertenecen a Volkswagen Group. Bentley fue adquirida en 1998 y actualmente se integra en el grupo de marcas dirigido por Audi.
- **Chrysler, Dodge y Maserati:** pertenecen actualmente a Stellantis, formado en 2021 mediante la unión de FCA y PSA. Previamente formaron parte de Fiat Chrysler Automobiles.
- **Mercedes-Benz y Maybach:** Maybach fue comercializada como una marca de lujo diferenciada entre 2002 y 2012. Desde 2014, Mercedes-Maybach constituye una submarca de Mercedes-Benz.
- **Chevrolet y General Motors:** Chevrolet continúa siendo una marca perteneciente a General Motors.

**3. Relaciones empresariales históricas**

- **Oldsmobile y General Motors:** Oldsmobile fue una marca de GM cuya producción terminó en 2004. Actualmente no se fabrican vehículos nuevos bajo esta marca.
- **Pontiac y General Motors:** Pontiac fue una marca de GM cuya comercialización terminó en 2010. Actualmente se encuentra descontinuada.
- **Hummer y General Motors:** GM comercializó vehículos Hummer desde 1999 hasta la desaparición de la antigua marca en 2010. Posteriormente, el nombre Hummer fue recuperado para vehículos eléctricos comercializados bajo GMC.
- **Saab y General Motors:** GM adquirió una participación del 50% en Saab Automobile en 1990 y asumió su control total en 2000. En 2010 la vendió y Saab Automobile terminó declarándose en quiebra en 2011. Actualmente la empresa Saab AB continúa operando en otros sectores, pero no produce automóviles Saab.
- **Mercedes-Benz y Chrysler:** pertenecieron al grupo DaimlerChrysler entre 1998 y 2007. Actualmente forman parte de grupos empresariales diferentes.
- **Aston Martin y Land Rover:** ambas pertenecieron a Ford durante el periodo 2000-2007. Aston Martin fue vendida en 2007 y Land Rover fue adquirida por Tata Motors en 2008. Actualmente no pertenecen al mismo grupo.

**4. Relaciones industriales y comerciales**

- **Giant Motors y FAW:** Giant Motors Latinoamérica es una ensambladora mexicana que ha producido vehículos comerciales de la marca FAW.

### 4. `Tipo`

- **Descripción:** identifica la denominación comercial del vehículo, que puede corresponder a un modelo comercial, una familia o gama de vehículos, o una versión o edición específica.
- **Tipo de dato original:** `object` (objeto de Python).
- **Naturaleza de la variable:** cualitativa nominal.
- **Valores distintos:** 220.
- **Valores faltantes:** 0.

**Análisis de las categorías observadas:**

Debido a la elevada cantidad de categorías distintas, se realizó una inspección exploratoria de las denominaciones más frecuentes y de una selección aleatoria de las restantes.

A partir de esta revisión, se identificaron los siguientes tipos de denominaciones:

- **Modelos comerciales:** nombres que identifican comercialmente un automóvil, como `A4`, `CHEVY` y `CORVETTE`.
- **Familias o gamas de vehículos:** denominaciones que agrupan vehículos de una misma línea comercial, los cuales pueden presentar diferencias de carrocería, configuración o características técnicas. Por ejemplo, `SERIE 1` de BMW y `SERIE C` de Mercedes-Benz.
- **Versiones o ediciones específicas:** denominaciones que identifican variantes particulares de un vehículo, diferenciadas por su equipamiento, características mecánicas o elementos estéticos. Por ejemplo, `CLASICO` y `ALLBLACK`, asociadas con vehículos MINI Cooper en los registros examinados.

Las clasificaciones anteriores no son necesariamente excluyentes, pues un modelo comercial también puede pertenecer a una familia o gama de vehículos.

Se identificó que la variable contiene denominaciones con diferentes niveles de especificidad, por lo que no presenta una clasificación completamente uniforme.

Finalmente, debido al carácter exploratorio de la inspección, no se garantiza que las 220 categorías correspondan exclusivamente a las clasificaciones identificadas.

### 12. `traccion`

- **Descripción:** indica la configuración del sistema de tracción del vehículo, especificando qué ruedas pueden recibir la fuerza motriz.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa nominal.
- **Valores distintos:** 5.
- **Categorías observadas:**
  - `4X2`: tracción en dos de las cuatro ruedas del vehículo.
  - `AWD`: sistema de tracción integral (*All-Wheel Drive*), que permite distribuir la potencia entre ambos ejes.
  - `2WD`: tracción en dos ruedas (*Two-Wheel Drive*).
  - `4X4`: configuración de tracción a las cuatro ruedas.
  - `4WD`: sistema de tracción a las cuatro ruedas (*Four-Wheel Drive*).
- **Valores faltantes:** 2,707 (94.39%)
**Observación:** se identifican categorías con significados equivalentes a nivel general, particularmente `4X2` y `2WD`, así como `4X4` y `4WD`. Estas equivalencias deberán considerarse durante la ingeniería de datos, conservando las posibles diferencias técnicas entre sistemas AWD y 4WD.
