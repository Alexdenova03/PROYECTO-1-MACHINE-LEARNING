# Diccionario de datos — Muestra 2

La Muestra 2 contiene **2,768 observaciones y 13 variables** que describen características comerciales y técnicas de los vehículos.

El presente diccionario tiene como objetivo documentar las características de cada variable, incluyendo su significado, tipo de dato, naturaleza estadística y valores observados.

## Descripción de las variables

### 1. `descripcion`

- **Descripción:** contiene una cadena de caracteres que reúne información comercial y técnica del vehículo, incluyendo su marca, denominación comercial y, dependiendo del registro, características mecánicas, de carrocería y de equipamiento.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa textual.
- **Valores distintos:** 772.
- **Valores faltantes:** 0.

**Análisis de la estructura de las descripciones**

Debido a la elevada cardinalidad de la variable, se realizó una inspección de las diez descripciones más frecuentes y de una selección aleatoria de 40 descripciones distintas, excluyendo las diez anteriores.

A partir de esta revisión, se identificó la siguiente estructura general:

**Marca o abreviatura → Denominación comercial → Características adicionales → Transmisión.**

Las características adicionales pueden incluir:

- **Versión o edición:** denominaciones particulares del vehículo, como `CARRERA 4S`, `PLATINUM`, `LUXURY`, `SAHARA`, `QUADRIFOGLIO`, `CARTIER` o `SIGNATURE`.
- **Carrocería y puertas:** características como `COUPE`, `CONVERTIBLE`, `CABRIOLET`, `SEDAN`, `SUV`, `ROADSTER`, `SPYDER`, así como el número de puertas (`2P`, `3P`, `4P`, `5P`).
- **Motor:** configuración, cilindrada y potencia, como `V6`, `V8`, `V10`, `V12`, `L4`, `L6`, `H4`, `H6`, `B4`, `3.0L`, `6.2L`, `230HP` o `423HP`.
- **Sobrealimentación:** denominaciones como `TURBO`, `BITBO`, `SUPERCARGADO`, `SC` o `ECOBOOST`.
- **Tracción:** denominaciones como `AWD`, `4X2`, `4X4`, `FWD` o `2WD`.
- **Equipamiento:** características como aire acondicionado (`AC`, `C/A`), quemacocos (`Q/C`, `QC`, `TCO`), vestiduras de piel o tela (`V/P`, `VP`, `V/T`), frenos (`ABS`), bolsas de aire (`BA`), sistemas de sonido o navegación (`CD`, `DVD`, `NAVI`), cámara trasera (`CAM TRA`) y medida de rines (`R18`, `R19`).
- **Sistemas de transmisión específicos y velocidades:** denominaciones como `PDK`, `TIPTRONIC`, `CVT`, `SELESPEED`, `TCT`, `SPORTSHIFT`, `SELECT SHIFT`, `6VEL` o `7SPEED`.
- **Transmisión (cierre):** denominaciones finales `AUT` y `STD`.

La cantidad y el orden de los elementos intermedios no son uniformes entre las descripciones. Algunos registros contienen únicamente la denominación comercial y la transmisión (por ejemplo, `LINCOLN LS STD.`), mientras que otros incluyen numerosas especificaciones adicionales.

**Identificación de las marcas**

Se extrajo la primera palabra de las 2,768 descripciones, identificándose 14 prefijos distintos. Posteriormente, se compararon dichos prefijos con la variable `Marca` para establecer sus correspondencias.

Se encontraron las siguientes equivalencias para los prefijos abreviados:

| Prefijo | Marca registrada | Patrón observado al inicio de `descripcion` |
|---|---|---|
| `AR` | ALFA ROMEO | Inicia con `AR ALFA ROMEO ...` en los 70 registros |
| `LX` | LEXUS | Inicia con `LX LEXUS ...` en los 5 registros |

Los 12 prefijos restantes (`BUICK`, `CADILLAC`, `INFINITI`, `JAC`, `JAGUAR`, `JEEP`, `LAMBORGHINI`, `LINCOLN`, `LOTUS`, `PORSCHE`, `SSANGYONG` y `SUBARU`) coinciden directamente y de forma biyectiva con la denominación registrada en `Marca`.

En el caso de `LEXUS`, se observó incluso un registro en el que el nombre de la marca se repite después del prefijo (`LX LEXUS LEXUS IS F SPORT V8 423HP AUT.`).

**Identificación de la denominación comercial**

Se observó que, después del prefijo inicial, generalmente aparece información que identifica comercialmente al vehículo. Sin embargo, al extraer la palabra posterior al prefijo y compararla con la variable `Tipo`, se obtuvieron **1,257 coincidencias exactas (45.41%)**. Las diferencias en el 54.59% restante se explican por los siguientes patrones:

- `PORSCHE 911 TURBO COUPE PDK AUT.`: la descripción presenta `911` después de la marca, mientras que `Tipo` repite la marca registrando `PORSCHE 911` (lo mismo ocurre en `BUICK ENVISION` e `INFINITI Q45`).
- `JEEP GRAND CHEROKEE SRT-8 4X4 AUT.`: la denominación comercial contiene varias palabras (`GRAND CHEROKEE`), mientras que `Tipo` agrupa bajo `CHEROKEE`.
- `CADILLAC CTS LUXURY...` y `LINCOLN MKZ SEDAN AUT.`: la descripción identifica el modelo específico (`CTS`, `MKZ`, `DEVILLE`, `BLS`), mientras que `Tipo` registra genéricamente la carrocería `SEDAN`.
- `LOTUS EXIGE S STD.` y `LX LEXUS LS 460L AUT.`: la descripción contiene el modelo (`EXIGE`, `EVORA`, `ELISE`, `LS 460L`, `RX 300`), pero `Tipo` repite únicamente el nombre de la marca (`LOTUS` y `LEXUS`).
- `AR ALFA ROMEO 156 4P V6 GTA STD.`: debido a que el prefijo abreviado va seguido del nombre compuesto de la marca (`AR ALFA ROMEO`), el modelo (`156`, `147`, `MITO`, `GIULIETTA`) se ubica a partir de la cuarta palabra.
- `LAMBORGHINI HURACAN LP 580-2 2P V10 5.2L AUT.`: presenta una inconsistencia directa en los datos, pues la descripción corresponde a un modelo `HURACAN`, mientras que `Tipo` registra `AVENTADOR`.

Por tanto, las descripciones presentan diferentes niveles de detalle y emplean nomenclaturas que no siempre coinciden literalmente con las de `Tipo`.

**Identificación de la transmisión**

Se verificaron las terminaciones de las 2,768 descripciones mediante la expresión regular `\b(?:AUT|STD)\.?\s*$`, comprobándose que **2,765 registros (99.89%)** finalizan directamente con alguna de las siguientes denominaciones:

- `AUT`: transmisión automática.
- `STD`: transmisión estándar o manual.

Los **3 registros restantes (0.11%)**, correspondientes a las descripciones `PORSCHE 911 CARRERA T COUPE 2P H6 3.0L STD .` (2 registros) y `PORSCHE 911 CARRERA T COUPE 2P H6 3.0L PDK AUT .` (1 registro), también finalizan indicando la transmisión (`STD` y `AUT`), pero presentan un espacio en blanco antes del punto final (`STD .` y `AUT .`). Considerando esta variación tipográfica, el 100% de las descripciones coincide con la columna `transmision`.

**Conclusión del análisis**

La variable `descripcion` contiene información vehicular estructurada parcialmente dentro de una cadena de texto. Aunque se identificaron patrones altamente consistentes en sus elementos iniciales y finales, el contenido intermedio presenta diferencias importantes en cuanto a nomenclatura, cantidad de información y nivel de especificidad, resguardando datos clave (como modelos específicos omitidos en `Tipo`, cilindradas, configuraciones de motor y tracción) que resultan fundamentales para el proceso de homologación.

### 2. `Ocupantes`

- **Descripción:** indica la capacidad de ocupantes o pasajeros del vehículo.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa nominal (con trasfondo cuantitativo discreto, al representar un conteo de plazas expresado como texto).
- **Valores distintos:** 6.
- **Valores faltantes:** 0.

**Categorías observadas:**

- `' 05 OCUP.'`: capacidad para cinco ocupantes (1,497 registros; 54.08%).
- `' 04 OCUP.'`: capacidad para cuatro ocupantes (520 registros; 18.79%).
- `' 02 OCUP.'`: capacidad para dos ocupantes (515 registros; 18.61%).
- `' 07 OCUP.'`: capacidad para siete ocupantes (188 registros; 6.79%).
- `' 08 OCUP.'`: capacidad para ocho ocupantes (40 registros; 1.45%).
- `' 06 OCUP.'`: capacidad para seis ocupantes (8 registros; 0.29%).

**Observación:** todas las categorías presentan un espacio en blanco al inicio de la cadena, numeración a dos dígitos con cero a la izquierda y la abreviatura `OCUP.`. En el catálogo asegurador (`Catalogo Proyecto 1.xlsx`), la variable equivalente `OCUPANTES` se encuentra codificada en formato numérico (`float64`).

### 3. `modelo`

- **Descripción:** representa el año-modelo del vehículo.
- **Tipo de dato original:** `int64` (entero).
- **Naturaleza de la variable:** cuantitativa discreta, con interpretación temporal.
- **Valores distintos:** 38.
- **Intervalo observado:** de 1983 a 2020.
- **Valores faltantes:** 0.

### 4. `Marca`

- **Descripción:** identifica la marca o denominación del fabricante del vehículo registrado en la muestra.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa nominal.
- **Valores distintos:** 14.
- **Valores faltantes:** 0.

**Categorías observadas:**

`ALFA ROMEO`, `BUICK`, `CADILLAC`, `INFINITI`, `JAC`, `JAGUAR`, `JEEP`, `LAMBORGHINI`, `LEXUS`, `LINCOLN`, `LOTUS`, `PORSCHE`, `SSANGYONG`, `SUBARU`.

**Análisis de equivalencias, relaciones comerciales y presencia en el catálogo**

La inspección de las categorías muestra que las 14 denominaciones corresponden a marcas automotrices distintas (sin duplicidad de abreviaturas dentro de la propia columna `Marca`), las cuales mantienen diversas relaciones corporativas e industriales:

**1. Relaciones empresariales vigentes en 2026**

- **Cadillac y Buick:** ambas marcas pertenecen a General Motors (GM).
- **Jeep y Alfa Romeo:** ambas pertenecen actualmente a Stellantis, grupo conformado en 2021 tras la fusión de Fiat Chrysler Automobiles (FCA) y Groupe PSA.
- **Porsche y Lamborghini:** ambas forman parte de Volkswagen Group. Lamborghini se integra operativamente como subsidiaria dentro del grupo de marcas dirigido por Audi.
- **Lincoln:** constituye la división de vehículos de lujo de Ford Motor Company.
- **Infiniti:** constituye la división de vehículos de lujo de Nissan Motor Co.
- **Lexus y Subaru:** Lexus es la división de vehículos de lujo de Toyota Motor Corporation; adicionalmente, Toyota mantiene una participación accionaria relevante y alianzas tecnológicas con Subaru Corporation.
- **Jaguar:** forma parte de JLR (Jaguar Land Rover), subsidiaria perteneciente al conglomerado indio Tata Motors desde 2008.
- **Lotus:** pertenece mayoritariamente al grupo automotriz chino Geely Holding Group desde 2017.
- **SsangYong:** tras su adquisición por KG Group en 2022, la compañía adoptó el nombre corporativo KG Mobility (KGM) en 2023.

**2. Relaciones empresariales históricas**

- **Jaguar y Lincoln:** ambas marcas coincidieron dentro de Ford Motor Company durante el periodo en que Jaguar formó parte del *Premier Automotive Group* de Ford (desde su adquisición en 1989/1999 hasta su venta a Tata Motors en 2008).

**3. Relaciones industriales y comerciales**

- **JAC:** en México, los vehículos de la marca de origen chino JAC (Jianghuai Automobile Co.) son ensamblados y comercializados por Giant Motors Latinoamérica (empresa presente en la Muestra 1).

**4. Presencia en el catálogo vehicular**

- De las 14 marcas registradas en la Muestra 2, **11 se encuentran directamente** en la variable `MARCA` del catálogo asegurador (`ALFA ROMEO`, `BUICK`, `CADILLAC`, `INFINITI`, `JAC`, `JAGUAR`, `JEEP`, `LEXUS`, `LINCOLN`, `PORSCHE` y `SUBARU`).
- Las **3 marcas restantes** (`LAMBORGHINI` con 44 registros, `LOTUS` con 23 registros y `SSANGYONG` con 4 registros) **no figuran** como categorías de la columna `MARCA` en el catálogo (únicamente existe un registro en el catálogo con `MARCA = 'GENERICO'`, `TIPO = 'GENERICO'` y `VERSION = 'SSANGYONG REXTON'`).

### 5. `Tipo`

- **Descripción:** identifica la denominación comercial del vehículo, que puede corresponder a un modelo comercial, una familia o gama de vehículos, una versión o edición específica, un tipo de carrocería o la repetición del nombre de la marca.
- **Tipo de dato original:** `object` (objeto de Python, al combinar valores enteros `int` y cadenas de caracteres `str`).
- **Naturaleza de la variable:** cualitativa nominal.
- **Valores distintos:** 98 (4 valores enteros: `147`, `156`, `159` y `166` asociados a Alfa Romeo; y 94 cadenas de texto).
- **Valores faltantes:** 0.

**Análisis de las categorías observadas:**

Debido a la cantidad de categorías distintas, se realizó una inspección exploratoria de las diez denominaciones más frecuentes (`PORSCHE 911`, `WRANGLER`, `SEDAN`, `CHEROKEE`, `ESCALADE`, `PORSCHE CAYENNE`, `PORSCHE BOXSTER`, `IMPREZA`, `PORSCHE PANAMERA` y `PATRIOT`), de una selección aleatoria de 20 categorías restantes y de las categorías ambiguas.

A partir de esta revisión, se identificaron los siguientes tipos de denominaciones:

- **Modelos comerciales directos:** nombres o códigos numéricos que identifican directamente un automóvil, como `WRANGLER`, `ESCALADE`, `IMPREZA`, `NAVIGATOR`, `FORESTER`, `Q50`, `GALLARDO`, `MURCIELAGO` o `159`.
- **Denominaciones compuestas por Marca + Modelo:** categorías que anteponen el nombre del fabricante al modelo comercial, tales como todas las categorías de Porsche (`PORSCHE 911`, `PORSCHE CAYENNE`, `PORSCHE BOXSTER`, `PORSCHE PANAMERA`, `PORSCHE CAYMAN`, `PORSCHE MACAN`, `PORSCHE 718 CAYMAN`, `PORSCHE 718 BOXSTER`), la mayoría de Buick (`BUICK ENCLAVE`, `BUICK ENCORE`, `BUICK REGAL`, `BUICK LACROSSE`, `BUICK VERANO`, `BUICK ENVISION`) y algunas de Infiniti (`INFINITI Q45`, `INFINITI I35`).
- **Familias o gamas que agrupan distintos modelos:** como `CHEROKEE` en Jeep, que agrupa tanto a vehículos `CHEROKEE` como `GRAND CHEROKEE`.
- **Denominaciones genéricas de carrocería (`SEDAN`):** en 223 registros (121 de `CADILLAC` y 102 de `LINCOLN`), la variable `Tipo` no registra el modelo comercial (`CTS`, `DEVILLE`, `SEVILLE`, `BLS`, `DTS`, `EL DORADO`, `TOWN CAR`, `MKZ`, `LS`, `CONTINENTAL`, e incluso el crossover `MKX`), sino únicamente la etiqueta genérica `SEDAN`.
- **Repetición del nombre de la marca (`LOTUS` y `LEXUS`):** en la totalidad de los registros de `LOTUS` (23) y `LEXUS` (5), la variable `Tipo` contiene nuevamente el nombre de la marca, omitiendo el modelo específico (`ELISE`, `EXIGE`, `EVORA`, `EUROPA`, `LS 460L`, `RX 300`, `IS F`).
- **Versiones o ediciones específicas:** denominaciones como `STERLING`, asociada con la edición especial `JEEP GRAND CHEROKEE STERLING 25 ANIVERSARIO 5P V6 3.6L 4X2 AUT.`, o `SUPER V8` en Jaguar.

Se concluye que la variable contiene denominaciones con diferentes niveles de especificidad y criterios de captura, además de presentar inconsistencias puntuales frente a la descripción (como los registros de `LAMBORGHINI HURACAN` etiquetados bajo el tipo `AVENTADOR`).

### 6. `AC`

- **Descripción:** indica si el vehículo cuenta con aire acondicionado.
- **Tipo de dato original:** `int64` (entero).
- **Naturaleza de la variable:** cualitativa nominal dicotómica (binaria).
- **Valores distintos:** 2.
- **Codificación:** 0 = No cuenta con aire acondicionado (2,086 registros; 75.36%); 1 = Cuenta con aire acondicionado (682 registros; 24.64%).
- **Valores faltantes:** 0.

### 9. `cilindrada`

- **Descripción:** representa la cilindrada del motor del vehículo, es decir, el volumen total desplazado por los pistones dentro de los cilindros, expresado en litros.
- **Tipo de dato original:** `float64` (número de punto flotante).
- **Naturaleza de la variable:** cuantitativa continua.
- **Valores distintos:** 28.
- **Valores observados:** se identificaron 28 valores distintos distribuidos dentro del intervalo de 1.4 a 6.2 litros (`1.4`, `1.5`, `1.6`, `1.8`, `2.0`, `2.2`, `2.3`, `2.4`, `2.5`, `2.7`, `2.9`, `3.0`, `3.2`, `3.4`, `3.5`, `3.6`, `3.7`, `3.8`, `4.0`, `4.2`, `4.8`, `5.0`, `5.2`, `5.3`, `5.4`, `5.7`, `6.0` y `6.2`).
- **Valores faltantes:** 1,570 (56.72%).

### 10. `motor`

- **Descripción:** indica la configuración del motor del vehículo, especificando la disposición y el número de cilindros.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa nominal.
- **Valores distintos:** 6.
- **Categorías observadas:**
  - `V6`: motor de seis cilindros dispuestos en V.
  - `L4`: motor de cuatro cilindros dispuestos en línea.
  - `V8`: motor de ocho cilindros dispuestos en V.
  - `V10`: motor de diez cilindros dispuestos en V.
  - `V12`: motor de doce cilindros dispuestos en V.
  - `L6`: motor de seis cilindros dispuestos en línea.
- **Valores faltantes:** 1,816 (65.61%).
**Observación:** en las descripciones textuales de vehículos Porsche y Subaru aparecen configuraciones adicionales de motores bóxer u horizontalmente opuestos (`H6`, `H4` y `B4`) que no figuran como categorías dentro de la columna `motor` (donde dichos registros suelen presentarse como valores faltantes o bajo otra categoría).

### 12. `transmision`

- **Descripción:** registra el tipo de transmisión del vehículo.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa nominal.
- **Valores distintos:** 2.
- **Categorías observadas:**
  - `AUT`: transmisión automática (2,172 registros; 78.47%).
  - `STD`: transmisión estándar o manual (596 registros; 21.53%).
- **Valores faltantes:** 0.
**Observación:** a diferencia de la Muestra 1, en la Muestra 2 esta variable se encuentra completamente estandarizada en las dos categorías básicas (`AUT` y `STD`), sin mezclar sistemas de tracción ni nombres comerciales de transmisiones (los cuales permanecen únicamente dentro de `descripcion`).

### 13. `traccion`

- **Descripción:** indica la configuración del sistema de tracción del vehículo, especificando qué ruedas pueden recibir la fuerza motriz.
- **Tipo de dato original:** `str` (cadena de caracteres).
- **Naturaleza de la variable:** cualitativa nominal.
- **Valores distintos:** 4.
- **Categorías observadas:**
  - `4X2`: tracción en dos de las cuatro ruedas del vehículo.
  - `AWD`: sistema de tracción integral (*All-Wheel Drive*), que permite distribuir la potencia entre ambos ejes.
  - `4X4`: configuración de tracción a las cuatro ruedas.
  - `2WD`: tracción en dos ruedas (*Two-Wheel Drive*).
- **Valores faltantes:** 1,907 (68.89%).
**Observación:** al igual que en la Muestra 1, se identifican categorías con significados equivalentes a nivel general (`4X2` y `2WD`), además de observarse en `descripcion` menciones adicionales como `FWD` (*Front-Wheel Drive*) en registros que tienen valor faltante en `traccion`.
